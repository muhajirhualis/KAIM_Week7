### Step 2: Test the Full Pipeline

```python
# Cell 3
question = "What are common issues with credit cards?"

# The ask() method runs the full RAG process: Retrieval -> Augmentation -> Generation
response = rag.ask(question)

print("="*60)
print(f"QUESTION: {response.question}")
print(f"ANSWER: {response.answer}")
print("="*60)

# Verify retrieved sources
print("\nTOP SOURCES RETRIEVED:")
for i, doc in enumerate(response.sources[:2], 1):
    # Use .get() to avoid KeyErrors if metadata is missing
    cid = doc.metadata.get('complaint_id', 'N/A')
    prod = doc.metadata.get('product', 'N/A')
    print(f"{i}. [ID: {cid}] {prod}: {doc.page_content[:150]}...")
```
